```python
def calculate_atr(experience: float, age: int) -> int:
    """
    Calculate the Annual Tax Revenue (ATR) based on years of experience and age.

    Rules:
    - If experience > 25 and age >= 55 → ATR = N5,600,000
    - Else if experience > 20 and age >= 45 → ATR = N4,480,000
    - Else if experience > 10 and age >= 35 → ATR = N1,500,000
    - Else → ATR = N550,000
    """
    if experience > 25 and age >= 55:
        return 5_600_000
    elif experience > 20 and age >= 45:
        return 4_480_000
    elif experience > 10 and age >= 35:
        return 1_500_000
    else:
        # Covers all other cases (including experience <= 10 or age < 35)
        return 550_000


def main():
    print("=" * 50)
    print("Izifin Technology - Annual Tax Revenue (ATR) Calculator")
    print("=" * 50)
    print("Enter staff details below. Type 'q' at any prompt to quit.\n")

    while True:
        try:
            # Get years of experience
            exp_input = input("Years of experience: ").strip()
            if exp_input.lower() == 'q':
                break
            experience = float(exp_input)
            if experience < 0:
                print("Experience cannot be negative. Please try again.\n")
                continue

            # Get age
            age_input = input("Age: ").strip()
            if age_input.lower() == 'q':
                break
            age = int(age_input)
            if age < 0:
                print("Age cannot be negative. Please try again.\n")
                continue

            # Calculate and display ATR
            atr = calculate_atr(experience, age)
            print(f"\n👉 Annual Tax Revenue (ATR): N{atr:,.0f}\n")
            print("-" * 30)

        except ValueError:
            print("Invalid input. Please enter numeric values.\n")
        except KeyboardInterrupt:
            print("\n\nExiting program.")
            break

    print("\nThank you for using the ATR Calculator.")


if __name__ == "__main__":
    main()
```
