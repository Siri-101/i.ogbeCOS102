```python
import tkinter as tk
from tkinter import messagebox

# Function to calculate ATR
def calculate_ATR():
    try:
        name = entry_name.get()
        age = int(entry_age.get())
        experience = int(entry_exp.get())

        if experience > 25 and age >= 55:
            atr = 5600000
        elif experience > 20 and age >= 45:
            atr = 4480000
        elif experience > 10 and age >= 35:
            atr = 1500000
        else:
            atr = 550000

        show_result(name, age, experience, atr)

    except ValueError:
        messagebox.showerror("Input Error", "Please enter valid numbers for Age and Experience.")


# Function to display result in a new window
def show_result(name, age, experience, atr):
    window = tk.Toplevel(root)
    window.title("ATR Result")
    window.geometry("400x250")

    label1 = tk.Label(window, text=f"Staff Name: {name}")
    label1.pack(pady=5)

    label2 = tk.Label(window, text=f"Age: {age}")
    label2.pack(pady=5)

    label3 = tk.Label(window, text=f"Years of Experience: {experience}")
    label3.pack(pady=5)

    label4 = tk.Label(
        window,
        text=f"Annual Tax Revenue (ATR): ₦{atr:,}",
        font=("Arial", 12, "bold")
    )
    label4.pack(pady=10)


# Main Window
root = tk.Tk()
root.title("Izifin Technology ATR Calculator")
root.geometry("400x300")

# Labels and Entry Widgets
label_name = tk.Label(root, text="Staff Name")
label_name.pack(pady=5)

entry_name = tk.Entry(root)
entry_name.pack(pady=5)

label_age = tk.Label(root, text="Age")
label_age.pack(pady=5)

entry_age = tk.Entry(root)
entry_age.pack(pady=5)

label_exp = tk.Label(root, text="Years of Experience")
label_exp.pack(pady=5)

entry_exp = tk.Entry(root)
entry_exp.pack(pady=5)

# Button
btn_calculate = tk.Button(root, text="Calculate ATR", command=calculate_ATR)
btn_calculate.pack(pady=20)

# Run Application
root.mainloop()
```
